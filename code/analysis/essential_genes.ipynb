# Count how many genes in our dataset are annotated as essential

```python
import pandas as pd
import numpy as np
from bson import decode_all
```

```python
df_wt = pd.read_csv("../../data/metadata/wt_sequences.csv")
df_wt.head()
```

```python
df = pd.read_csv("../../data/metadata/essential_genes.csv", header=1)
df = df[df['Essential'] == True]
df['Gene']
```

```python
df[[np.any([y in x for x in df_wt.genes]) for y in df['Gene']]]
```

Additionally we can check which transcription factors are essential.

```python
df_TFs = pd.read_csv("../../data/metadata/all_tfs.csv")
df_TFs
```

```python
df_TF_alt = df_TFs.dropna()
df_TF_alt
```

```python
pd.merge(df_TFs, df, right_on='Gene', left_on='Gene Name', how='inner')
```

```python
pd.merge(df_TF_alt, df, right_on='Gene', left_on='Previous/Alternative Name', how='inner')
```
